```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

```python
# read parrot.png
parrot = Image.open('parrot.png')
plt.imshow(parrot)
plt.axis('off')
plt.show()

print('Image size: ', parrot.size)
```

```python
# convert to numpy
I = np.array(parrot).astype(float)  # avoid uint8 overflow in later arithmetic

# let us compute the dimension of the vector space:
print('Shape of image:')
print(I.shape)
print('Number of pixels (dimension of space): ')
dim = I.shape[0] * I.shape[1] 
print(dim)
```

```python
# define a convolution kernel, to compute the 
# average of a square portion of the image

m = 21

def simple_blur_kernel(m):
    """ A simple average kernel, averages an m x m block """
    A = np.ones((m, m)) / (m * m)
    return A


def gaussian_edge_kernel(m, sigma=None, axis=0):
    """
    First derivative of a Gaussian, for odd m.
    axis=0 -> detects horizontal edges (d/dy)
    axis=1 -> detects vertical edges   (d/dx)
    """
    assert m % 2 == 1
    if sigma is None:
        sigma = m / 6

    r = np.arange(-(m//2), m//2 + 1)
    x, y = np.meshgrid(r, r)

    G = np.exp(-(x*x + y*y) / (2*sigma*sigma))

    if axis == 0:
        K = -y / sigma**2 * G
    else:
        K = -x / sigma**2 * G

    # Normalize scale; derivative kernel already sums ~0
    K /= np.sum(np.abs(K))
    return K

#A = simple_blur_kernel(21)
A = gaussian_edge_kernel(21, axis=1)
```

```python
I_blur = np.zeros((I.shape[0]-m, I.shape[1]-m))

for i in range(I.shape[0]-m):
    for j in range(I.shape[1]-m):
        
        I_blur[i, j] = np.sum(A * I[i:i+m, j:j+m]) / (m * m)
        
        
```

```python
plt.figure()
plt.imshow(I_blur, cmap='gray')
plt.show()
```

```python
# We wrap the blurring into a function

def blur(I, m=21):
    "Blur the image I with an m x m kernel"
    A = simple_blur_kernel(m)
    I2 = np.zeros((I.shape[0]-m, I.shape[1]-m))
    for i in range(I.shape[0]-m):
        for j in range(I.shape[1]-m):
            I2[i, j] = np.sum(A * I[i:i+m, j:j+m])
    return I2
```

```python
# Make a flipped image
FI = np.flipud(I)

# Blur it
BFI = blur(FI)

plt.figure()
plt.imshow(FI, cmap='gray')
plt.show()

plt.figure()
plt.imshow(BFI, cmap='gray')
plt.show()
```

```python
# Let us take linear combination _then_ blur:
# and blur _then_ linear combination

BI = blur(I)
X = 0.5 * (BI + BFI)
Y = blur(0.5 * (I + FI))


plt.figure()
plt.imshow(X, cmap = 'gray')
plt.show()

plt.figure()
plt.imshow(Y, cmap = 'gray')
plt.show()
```

```python
# compare with linear combination of blurred images
plt.figure()

plt.imshow(X - Y)
plt.colorbar()
```
