```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
from PIL import Image
```

```python
# read parrot.png
parrot = Image.open('parrot.png').convert('L')
P = np.array(parrot, dtype=float) 
print(P.shape)
plt.imshow(P, cmap='gray')
plt.axis('off')
plt.show()

print('Image size: ', parrot.size)
```

```python
U, S, V = np.linalg.svd(P)
```

```python
print(f'Shape of U: {U.shape}, length of S: {len(S)}, shape of V: {V.shape}')
```

```python
plt.figure()
plt.imshow(np.log10(np.abs(U)),vmax=0,vmin=-3,cmap='viridis')
plt.colorbar()
plt.title('Log10 of abs of elements of U')
plt.show()


plt.figure()
plt.imshow(np.log10(np.abs(V)),vmax=0,vmin=-3,cmap='viridis')
plt.colorbar()
plt.title('Log10 of abs of elements of V')
plt.show()
```

```python
plt.figure()
plt.semilogy(S)
plt.ylabel(r'$\log \sigma_i$')
```

```python
def approx_image(nmax):
    """ Constsruct an approximation to the image by including only a portion of singular values """
    return U[:,:nmax] @ np.diag(S[:nmax]) @ V[:nmax,:]
```

```python
plt.figure()
plt.imshow(approx_image(1),cmap='gray')
```
