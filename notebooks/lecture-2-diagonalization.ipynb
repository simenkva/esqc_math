### The eigenvalue problem

We seek the stationary states of the vibrational Hamiltonian

$$
\hat H\psi_k(r)=E_k\psi_k(r),
\qquad
\hat H=-\frac{1}{2\mu}\frac{d^2}{dr^2}+V(r).
$$

After finite-difference discretization, this becomes the Hermitian matrix eigenvalue problem

$$
A\mathbf{x}_k=\lambda_k\mathbf{x}_k.
$$

For a Hermitian matrix, `lamb, X = np.linalg.eigh(A)` returns the eigenvalues in ascending order in `lamb` and the corresponding normalized eigenvectors as **columns** of `X`:

$$
\lambda_k = \mathrm{lamb}[k],
\qquad
\mathbf{x}_k = X[:,k],
$$

so that

$$
H\,X[:,k] = \mathrm{lamb}[k]\,X[:,k].
$$

Here $\lambda_k\approx E_k$, and $\mathbf{x}_k$ represents the discretized vibrational wavefunction $\psi_k(r)$ on the grid points.

```python
import numpy as np
import matplotlib.pyplot as plt
from plotsetup import *
```

```python
def morse(r, De=0.225, a=1.174, re=1.733):
    """Morse potential for HF in atomic units."""
    return De * (1 - np.exp(-a * (r - re)))**2 - De

mu = 1741.1 # reduced mass of HF

n = 100
r = np.linspace(1, 6, 100)
fig, ax = plt.subplots(figsize=(6,3))
plt.plot(r, morse(r), label='V(r) = Morse potential for HF')
plt.legend()
plt.xlabel('r')
plt.ylabel('V(r)')
plt.grid()
plt.show()
```

```python
# set up a finite difference discretization of 
#
# H = -1/2 d^2/dr^2 + V(r)
#
from scipy.sparse import diags

dr = r[1]-r[0]
T = diags([-1.0, 2.0, -1.0], [-1, 0, 1], shape=(n, n)) / (2*mu*dr**2)
V = np.diag(morse(r))
A = T + V
```

```python
plt.figure()
plt.imshow(A)
plt.colorbar()
plt.show()
```

```python
# Is H hermitian?

print('Upper left corner of A:')
print(A[:3,:3])

print('max abs difference of A and A^H:')
print(np.max(np.abs(A - A.conj().T)))
```

```python
# Compute eigenvectors and eigenvalues

lamb, X = np.linalg.eigh(A)
```

```python
# the first few eigenvalues

print(lamb[:5])
```

```python
plt.figure()
plt.imshow(X[:,:])
plt.show()
```

```python
# plot eigenvectors


ax.plot(r, lamb[0] + .2 * X[:,0], color = 'C1', label=r'$\psi_0$')
ax.axhline(lamb[0], color='C1', linestyle='--', linewidth=.5)

ax.plot(r, lamb[1] + .2 * X[:,1], color = 'C1', label=r'$\psi_2$')
ax.axhline(lamb[1], color='C1', linestyle='--', linewidth=.5)


ax.legend()
fig
```

```python
# Check spectral decomposition

A2 = X @ np.diag(lamb) @ X.conj().T
```

```python
plt.figure()
plt.imshow(A2 - A)
plt.colorbar()
plt.show()
```

```python
# Show X is unitary

I = X.conj().T @ X

plt.figure()
plt.imshow(I)
plt.colorbar()
plt.show()
```
