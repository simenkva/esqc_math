```python
import numpy as np
import matplotlib.pyplot as plt

from plotsetup import *
```

```python
fig, ax = new_plot()

# define two points

p = np.array([1, 2])
q = np.array([2, -3])

draw_point(ax, p, color='red')
draw_vector(ax, p, color='red')
draw_point(ax, q, color='blue')
draw_vector(ax, q, color='blue')
```

```python
# vector sum
r = p + q
print('p = ', p)
print('q = ', q)
print('p + q = ', p + q)

draw_point(ax, r, color='green')
draw_vector(ax, r, color='green')
fig
```

```python
draw_vector(ax, p, origin=q, color='black')
draw_vector(ax, q, origin=p, color='black')
fig
```

```python
# make a line

fig, ax = new_plot()

alpha_range = np.linspace(-1, 1, 10)
for alpha in alpha_range:
    
    z = p + alpha * q
    
    draw_point(ax, z, color='blue')
    draw_vector(ax, z, color='blue')
    

```

```python
# norm and inner product

print('p = ', p)
print('q = ', q)

# compute inner product by formula
print(p[0]*q[0] + p[1]*q[1])

# by numpy function
print(np.dot(p, q))
```

```python
# compute norms of vectors
norm_p = np.sqrt(np.dot(p,p))
norm_q = np.sqrt(np.dot(q,q))

# compute angle
angle = np.arccos(np.dot(p,q)/norm_p / norm_q)
print(angle)
```

```python
# convert to degrees
print(180 * angle/np.pi)
```
