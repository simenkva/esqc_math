```python
import numpy as np
import matplotlib.pyplot as plt

from plotsetup import *

```

```python
fig, ax = new_plot()

# define two points

x = np.array([2, 1])
y = np.array([1, 2])

draw_point(ax, x, color='red')
draw_vector(ax, x, color='red')
draw_point(ax, y, color='blue')
draw_vector(ax, y, color='blue')

fig
```

```python
# vector sum
z = x + y
print(x, y, z)

draw_point(ax, z, color='green')
draw_vector(ax, z, color='green')
```

```python
draw_vector(ax, x, origin=y, color='black')
fig
```

```python
# make a line

fig, ax = new_plot()

alpha_range = np.linspace(-1, 1, 10)
for alpha in alpha_range:
    
    z = x + alpha * y
    
    draw_point(ax, z, color='blue')
    draw_vector(ax, z, color='blue')
    

```

```python
# norm and inner product

print('x = ', x)
print('y = ', y)

# compute inner product by formula
print(x[0]*y[0] + x[1]*y[1])

# by numpy function
print(np.dot(x, y))
```

```python
# compute norms of vectors
norm_x = np.sqrt(np.dot(x,x))
norm_y = np.sqrt(np.dot(y,y))

# compute angle
angle = np.arccos(np.dot(x,y)/norm_x / norm_y)
print(angle)
```

```python
# convert to degrees
print(180 * angle/np.pi)
```
