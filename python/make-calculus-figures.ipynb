```python
import numpy as np
import matplotlib.pyplot as plt
from plotsetup import *

# increase font sizes
plt.rcParams.update({'font.size': 16})
```

```python
fig, ax = new_plot()

x = np.linspace(-4, 4, 400)
y = 1/x
ax.plot(x, y)
# add annotation 
ax.text(2, 2, r'$f(x) = \frac{1}{x}$', fontsize=24)
plt.savefig('1_over_x.svg')
plt.show()
```

```python
fig, ax = new_plot(xlim=(-.1, 1), ylim=(-1.5, 1.5))

x = np.linspace(1e-10, 1, 4000)
f = np.sin(1/np.abs(x))
ax.plot(x, f)
# turn off grid
ax.grid(False)

# add annotation text
ax.text(0.5, 1.2, r'$f(x) = \sin\left(\frac{1}{|x|}\right)$', fontsize=24)
plt.savefig('sin_1_over_x_squared.svg')
plt.show()
```

```python
fig, ax = new_plot()

x1 = np.linspace(-4, 1, 400)[:-1]
f1 = x1 + (x1>=1)
ax.plot(x1, f1)
x2 = np.linspace(1, 4, 400)
f2 = x2 + (x2>=1)
ax.plot(x2, f2, color='C0')
# add circle
circle = plt.Circle((1, 2), 0.1, color='C0', fill=True)
ax.add_patch(circle)
# turn off grid
ax.grid(False)

# add annotation text
#ax.text(-3.5, 1.2, r'$f(x) = x + (x\geq1)$', fontsize=16)
plt.savefig('x_plus_step_function.svg')
plt.show()
```

```python
fig, ax = new_plot()
f = lambda x: 0.5*x**3 - x+ 2
df = lambda x: 1.5*x**2 - 1
x0 = 1
y0 = f(x0)
dy0 = df(x0)
x = np.linspace(-3, 3, 400)
ax.plot(x, f(x))
ax.plot(x0, y0, 'ro')  # mark the point (x0, y0)
ax.plot(x, y0 + dy0*(x - x0), 'C1-')  # plot the tangent line at x0
plt.savefig('derivative.svg')
plt.show()
```
